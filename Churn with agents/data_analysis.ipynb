## Categorical features

```python
cat = df.select_dtypes(include='object').columns[1:]
```

```python
for i in cat:
    print(df[i].value_counts())
```

```python
for i in cat:
    df_out = df[[i,'Exited']].groupby(i).agg( ['sum','count']).reset_index()
    df_out.columns = ['_'.join(col).strip('_') for col in df_out.columns.values]
    df_out['churn_share'] = df_out['Exited_sum'] / df_out['Exited_count']
    plot1= px.bar(df_out,x=i, y = 'churn_share')
    plot1.show()
    
```

## Deep learning

```python
import keras as kr
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
df_new_scaled = sc.fit_transform(df_new[features_2],)
```

```python
train_x_dl = df_new_scaled[train_x.index]
test_x_dl = df_new_scaled[test_x.index]
```

```python
model = kr.models.Sequential([
    kr.layers.Dense(64,'relu',input_shape = (6,)),
    kr.layers.Dense(32,'relu'),
    kr.layers.Dense(1,'sigmoid')
])
```

```python
model.compile(kr.optimizers.Adam(),
              loss = kr.losses.BinaryCrossentropy(),
              metrics=[kr.metrics.AUC()])
```

```python
m_dl = model.fit(train_x_dl, train_y,
                 validation_data=(test_x_dl, test_y),
                 epochs=300,
                 batch_size=20)
```
